```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import rasterio
import h5py

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

print("All libraries imported successfully!")
```

```python
height = 100
width = 100
bands = 30

hyperspectral_cube = np.random.rand(
    height, width, bands
).astype(np.float32)

print("Shape:", hyperspectral_cube.shape)
print("Height:", height)
print("Width:", width)
print("Spectral Bands:", bands)
print("Data Type:", hyperspectral_cube.dtype)
```

```python
plt.figure(figsize=(7, 5))

plt.imshow(hyperspectral_cube[:, :, 10], cmap="viridis")

plt.colorbar(label="Pixel Intensity")
plt.title("TerraSpectra - Sample Spectral Band 10")
plt.xlabel("Width")
plt.ylabel("Height")

plt.show()
```

```python
pixel_spectrum = hyperspectral_cube[50, 50, :]

plt.figure(figsize=(9, 4))

plt.plot(range(1, 31), pixel_spectrum)

plt.title("Spectral Signature of a Sample Pixel")
plt.xlabel("Spectral Band")
plt.ylabel("Reflectance Value")
plt.grid(True)

plt.show()
```
